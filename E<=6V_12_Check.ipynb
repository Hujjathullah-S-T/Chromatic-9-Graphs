```python
import pandas as pd

# CSV file
file = "/content/generated_9_chromatic_graphs_test(10) (2).csv"

df = pd.read_csv(file)

# Calculate bound
df["6V-12"] = 6 * df["n_nodes"] - 12

# Check condition
df["E <= 6V-12"] = df["n_edges"] <= df["6V-12"]

# Display results
print(df[
    [
        "generated_id",
        "n_nodes",
        "n_edges",
        "6V-12",
        "E <= 6V-12"
    ]
].to_string(index=False))

# Summary
print("\n======================================")
print("SUMMARY")
print("======================================")

print("Total graphs :", len(df))
print(
    "E <= 6V-12   :",
    df["E <= 6V-12"].sum()
)
print(
    "E > 6V-12    :",
    (~df["E <= 6V-12"]).sum()
)

# Save results
df[
    [
        "generated_id",
        "n_nodes",
        "n_edges",
        "6V-12",
        "E <= 6V-12"
    ]
].to_csv(
    "E_leq_6V_minus_12_results.csv",
    index=False
)

print("\nSaved as: E_leq_6V_minus_12_results.csv")
```
